```python
ALL_EXCLUDED_PIDS = set(FAILED_ATTENTION_PIDS + AI_PIDS)
```

```python
from scipy import stats
from matplotlib.patches import Patch

old_trials_df = df.query('old_trial == 1').dropna(subset=['optimal_old_choice']).copy()

# One-sample binomial test per subject against chance (0.5), one-tailed
def subject_binom_test(g):
    k = int(g['optimal_old_choice'].sum())
    n = len(g['optimal_old_choice'])
    p = stats.binomtest(k, n, 0.5, alternative='greater').pvalue
    return pd.Series({'pval': p, 'significant': p < 0.05})

sig_test_df = old_trials_df.groupby('participant_id').apply(subject_binom_test, include_groups=False).reset_index()

plot_df = old_trials_df.groupby('participant_id', as_index=False).agg({'optimal_old_choice': 'mean'})
plot_df = plot_df.merge(sig_test_df, on='participant_id')

# Sort: non-significant first, then significant; within each group sort by performance
plot_df['sig_order'] = plot_df['significant'].astype(int)
plot_df = plot_df.sort_values(['sig_order', 'optimal_old_choice']).reset_index(drop=True)
plot_df['x'] = np.arange(len(plot_df))

def overall_bar_color(row):
    if row['participant_id'] in AI_PIDS:
        return 'mediumpurple'
    elif row['participant_id'] in FAILED_ATTENTION_PIDS:
        return 'tomato'
    elif row['significant']:
        return 'steelblue'
    return 'gray'

colors = [overall_bar_color(row) for _, row in plot_df.iterrows()]

fig, ax = plt.subplots(figsize=(12, 4))
ax.bar(plot_df['x'], plot_df['optimal_old_choice'], color=colors)
ax.set_title(f"Overall Old Trial Optimal Performance (n={plot_df['participant_id'].nunique()})")
ax.set_xlabel('Participant')
ax.set_ylabel('Optimal old choice rate')
ax.set_xticks(plot_df['x'])
ax.set_xticklabels(plot_df['participant_id'], rotation=30, ha='right', fontsize=3)
ax.axhline(y=0.5, linestyle='dashed', color='red')
ax.set_ylim(0, 1)

# Vertical line between non-significant and significant participants
n_not_sig = int((~plot_df['significant']).sum())
if 0 < n_not_sig < len(plot_df):
    ax.axvline(x=n_not_sig - 0.5, color='black', linestyle='--', linewidth=1.5)

n_sig = int(plot_df['significant'].sum())
n_total = len(plot_df)
print(f'{n_sig} out of {n_total} participants significantly above chance (one-sample binomial test, p < 0.05, one-tailed)')

legend_elements = [
    Patch(facecolor='steelblue', label='Significant above chance'),
    Patch(facecolor='lightgray', label='Not significant'),
    Patch(facecolor='tomato', label='Failed attention check'),
    Patch(facecolor='mediumpurple', label='Suspected AI'),
]
ax.legend(handles=legend_elements, loc='upper left', fontsize=8)

plt.tight_layout()
if save:
    plt.savefig(os.path.join(figures_dir, 'optimal_subject.png'))
plt.show()

# save significant participants
sig_df = plot_df[plot_df['significant']]
sig_clean_df = df[df['participant_id'].isin(sig_df['participant_id']) & ~df['participant_id'].isin(ALL_EXCLUDED_PIDS)].query('old_trial == 1').dropna(subset=['optimal_old_choice'])
```

# Mixed memorability analysis

df.ret_type:
- 1 -> high=0, low=1
- 2 -> high=1, low=0
- 3 -> high=0, low=0
- 4 -> high=1, low=1

```python
## Setup 
INCOMPLETE_PIDS = df[df.old_trial==1].groupby('participant_id').size().loc[lambda s: s < 70].index.tolist()
ALL_EXCLUDED_PIDS = set(FAILED_ATTENTION_PIDS + AI_PIDS + INCOMPLETE_PIDS)

old_mixed = df[~df['participant_id'].isin(AI_PIDS)].query('old_trial == 1').copy()
old_mixed['optimal_old_choice'] = pd.to_numeric(old_mixed['optimal_old_choice'], errors='coerce')
# Which memorability bin did the participant effectively choose?
old_mixed['chose_high_mem'] = (
    ((old_mixed['chosen_side'] == 'left')  & (old_mixed['left_mem_bin']  == 'high')) |
    ((old_mixed['chosen_side'] == 'right') & (old_mixed['right_mem_bin'] == 'high'))
).astype(float)

# Human-readable labels & display order
RT_LABELS = {1: 'high=$0\nlow=$1', 2: 'high=$1\nlow=$0', 3: 'both=$0', 4: 'both=$1'}
RT_ORDER  = ['high=$0\nlow=$1', 'high=$1\nlow=$0', 'both=$0', 'both=$1']
old_mixed['ret_type_label'] = old_mixed['ret_type'].map(RT_LABELS)

even_df   = old_mixed[old_mixed['ret_type'].isin([3, 4])].copy()
uneven_df = old_mixed.dropna(subset=['optimal_old_choice']).copy()  

# Significant participants: one-sample binomial test on uneven types, p < 0.05 one-tailed
def _binom_test_mixed(g):
    vals = g['optimal_old_choice'].dropna()
    k = int(vals.sum())
    n = len(vals)
    p = stats.binomtest(k, n, 0.5, alternative='greater').pvalue
    return pd.Series({'pval': p, 'significant': p < 0.05})

sig_mixed_summary = uneven_df.groupby('participant_id').apply(_binom_test_mixed, include_groups=False).reset_index()
sig_mixed_pids = sig_mixed_summary.query('significant')['participant_id'].tolist()
print(f'{len(sig_mixed_pids)} / {df.participant_id.nunique()} participants significantly above chance on uneven trials')

sig_clean_mixed = old_mixed[
    old_mixed['participant_id'].isin(sig_mixed_pids) &
    ~old_mixed['participant_id'].isin(ALL_EXCLUDED_PIDS)
].copy()
print(f'Included n = {sig_clean_mixed.participant_id.nunique()}')
```

# Matched-memorability analysis

```python
## Setup
INCOMPLETE_PIDS = df[df.old_trial==1].groupby('participant_id').size().loc[lambda s: s < 70].index.tolist()
ALL_EXCLUDED_PIDS = set(FAILED_ATTENTION_PIDS + AI_PIDS + INCOMPLETE_PIDS)

old_matched = df[~df['participant_id'].isin(AI_PIDS)].query('old_trial == 1').copy()
old_matched['optimal_old_choice'] = pd.to_numeric(old_matched['optimal_old_choice'], errors='coerce')
old_matched['chosen_value'] = pd.to_numeric(old_matched['chosen_value'], errors='coerce')

MEM_ORDER = ['low', 'high']

# Significant participants: one-sample binomial test on old trials, p < 0.05 one-tailed
def _binom_test_matched(g):
    vals = g['optimal_old_choice'].dropna()
    k = int(vals.sum())
    n = len(vals)
    p = stats.binomtest(k, n, 0.5, alternative='greater').pvalue
    return pd.Series({'pval': p, 'significant': p < 0.05})

sig_matched_summary = old_matched.dropna(subset=['optimal_old_choice']).groupby('participant_id').apply(_binom_test_matched, include_groups=False).reset_index()
sig_matched_pids = sig_matched_summary.query('significant')['participant_id'].tolist()
print(f'{len(sig_matched_pids)} / {df.participant_id.nunique()} participants significantly above chance on old trials')

sig_clean_matched = old_matched[
    old_matched['participant_id'].isin(sig_matched_pids) &
    ~old_matched['participant_id'].isin(ALL_EXCLUDED_PIDS)
].copy()
print(f'Included n = {sig_clean_matched.participant_id.nunique()}')
```

# Direct-value memory test analysis

```python
from scipy import stats

# Direct task has no old/new choice on repeated-card trials -- it has two
# separate memory tests instead: recognition (did you pick the seen-before
# card?) and a value report (what was it worth?). Build per-test dataframes
# and use recognition accuracy (analogous to optimal_old_choice elsewhere)
# to define the per-subject significance / inclusion set.
MEM_ORDER = ['low', 'mid', 'high']

recog_df = df[~df['participant_id'].isin(AI_PIDS)].query('is_recognition_trial == True').copy()
recog_df['recognition_correct'] = pd.to_numeric(recog_df['recognition_correct'], errors='coerce')
recog_df['choice_missed'] = pd.to_numeric(recog_df['choice_missed'], errors='coerce')

value_df = df[~df['participant_id'].isin(AI_PIDS)].query('is_value_test_trial == True').copy()
value_df['value_test_correct'] = pd.to_numeric(value_df['value_test_correct'], errors='coerce')
value_df['value_test_missed'] = pd.to_numeric(value_df['value_test_missed'], errors='coerce')
value_df['old_value'] = pd.to_numeric(value_df['old_value'], errors='coerce')

INCOMPLETE_PIDS = (
    df[df['is_recognition_trial'] == True]
    .groupby('participant_id').size()
    .loc[lambda s: s < 100]
    .index.tolist()
)

# A subset of sessions (mostly Windows/Chrome) have a bug where keyboard
# responses on the value-report screen intermittently fail to register.
# Exclude anyone with > 20% missed trials on either recognition or value-report.
recog_miss_by_subj = recog_df.groupby('participant_id')['choice_missed'].mean()
value_miss_by_subj = value_df.groupby('participant_id')['value_test_missed'].mean()
HIGH_MISS_PIDS = list(set(
    recog_miss_by_subj[recog_miss_by_subj > 0.2].index.tolist() +
    value_miss_by_subj[value_miss_by_subj > 0.2].index.tolist()
))
print(f'{len(HIGH_MISS_PIDS)} participants excluded for > 20% miss rate on recognition or value-report trials')

ALL_EXCLUDED_PIDS_DIRECT = set(FAILED_ATTENTION_PIDS + AI_PIDS + INCOMPLETE_PIDS + HIGH_MISS_PIDS)

# One-sample binomial test per subject against chance (0.5) on recognition accuracy, one-tailed
def _binom_test_recognition(g):
    vals = g['recognition_correct'].dropna()
    k = int(vals.sum())
    n = len(vals)
    p = stats.binomtest(k, n, 0.5, alternative='greater').pvalue
    return pd.Series({'pval': p, 'significant': p < 0.05})

sig_recog_summary = (
    recog_df.dropna(subset=['recognition_correct'])
    .groupby('participant_id').apply(_binom_test_recognition, include_groups=False)
    .reset_index()
)
sig_recog_pids = sig_recog_summary.query('significant')['participant_id'].tolist()
print(f'{len(sig_recog_pids)} / {df.participant_id.nunique()} participants significantly above chance on recognition')

sig_clean_recog = recog_df[
    recog_df['participant_id'].isin(sig_recog_pids) &
    ~recog_df['participant_id'].isin(ALL_EXCLUDED_PIDS_DIRECT)
].copy()
sig_clean_value = value_df[
    value_df['participant_id'].isin(sig_recog_pids) &
    ~value_df['participant_id'].isin(ALL_EXCLUDED_PIDS_DIRECT)
].copy()
print(f'Included n = {sig_clean_recog.participant_id.nunique()}')
```
